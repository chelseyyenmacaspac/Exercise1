## Step 1: Classify the Customer Issue

Prompt:

You are a customer support AI. Read the customer's message and classify the issue into one of these categories: Billing, Technical Issue, Account Access, Order/Delivery, or Other.

Customer message: "I was charged twice for my monthly subscription."

Respond with the category and a one-sentence explanation.

```python
customer_message = "I was charged twice for my monthly subscription."

print("Customer Message:")
print(customer_message)

classification = "Billing - The customer was charged twice for the same subscription."

print("\nStep 1 Output:")
print(classification)
```

## Step 2: Gather Missing Information

Prompt:

You are a customer support AI. Based on the issue classification from Step 1, determine what additional information is needed from the customer before solving the problem.

Ask only 2 relevant questions. Keep the questions polite and simple.

Use the classification from Step 1 as your input.

```python
prompt2 = f"""
You are a customer support AI.

The issue was classified in Step 1 as:
{classification}

Based on this classification, determine what additional information
is needed from the customer before solving the problem.

Ask only 2 relevant questions.
Keep the questions polite and simple.
"""

print("Step 2 Prompt:")
print(prompt2)

missing_info = """
1. What date were the two charges made?
2. Can you confirm the amount of each charge?
"""

print("\nStep 2 Output:")
print(missing_info)
```

## Step 3: Propose a Solution

Prompt:

You are a customer support AI. Based on the issue classification and the information requested in the previous steps, propose a solution for the customer.

Give a clear and helpful response in 2–3 sentences. Keep the tone polite and professional.

```python
prompt3 = f"""
You are a customer support AI.

Issue classification from Step 1:
{classification}

Information requested in Step 2:
{missing_info}

Based on the previous steps, propose a solution for the customer.

Give a clear and helpful response in 2-3 sentences.
Keep the tone polite and professional.
"""

print("Step 3 Prompt:")
print(prompt3)

solution = """
Once the duplicate charges are confirmed, the billing team can review
the transactions and refund the extra charge if it was made in error.
The customer should provide the charge dates and amounts for verification.
"""

print("\nStep 3 Output:")
print(solution)
```

## Step 4: Decide Whether to Escalate

Prompt:

You are a customer support AI. Review the issue classification and proposed solution from the previous steps.

Decide whether the issue should be escalated to a human customer service agent.

Respond with "Escalate" or "Do Not Escalate" and give a one-sentence explanation.

```python
prompt4 = f"""
You are a customer support AI.

Issue classification from Step 1:
{classification}

Proposed solution from Step 3:
{solution}

Decide whether this issue should be escalated to a human customer
service agent.

Respond with "Escalate" or "Do Not Escalate" and give a
one-sentence explanation.
"""

print("Step 4 Prompt:")
print(prompt4)

escalation = """
Escalate - A human billing agent should verify the duplicate charges
and process a refund if necessary.
"""

print("\nStep 4 Output:")
print(escalation)
```

```python
improved_prompt = f"""
You are a customer support AI.

Issue classification:
{classification}

Proposed solution:
{solution}

Decide whether the issue should be escalated to a human agent.

Escalate if:
- A refund needs to be processed
- An account change requires human approval
- The issue cannot be solved automatically

Respond with "Escalate" or "Do Not Escalate"
and give a one-sentence explanation.
"""

print("Improved Prompt:")
print(improved_prompt)

improved_output = """
Escalate - The issue requires a human billing agent because a refund
may need to be processed for the duplicate charge.
"""

print("\nImproved Output:")
print(improved_output)
```

## Final Successful Output

The following shows the complete customer support prompt chain. Each step uses information from the previous step to classify the problem, gather information, propose a solution, and determine whether human escalation is needed.


```python
print("=== CUSTOMER SUPPORT AI - FINAL OUTPUT ===")

print("\nCustomer Message:")
print(customer_message)

print("\nStep 1 - Issue Classification:")
print(classification)

print("\nStep 2 - Missing Information:")
print(missing_info)

print("\nStep 3 - Proposed Solution:")
print(solution)

print("\nStep 4 - Escalation Decision:")
print(improved_output)
```
